```python
from sklearn import svm
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.decomposition import PCA
from seaborn import heatmap
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("titanic.csv")
le = LabelEncoder()

for col in df:
    df[col] = le.fit_transform(df[col])

x = df[['PassengerId', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked']]
y = df["Survived"]
```

```python
mms = MinMaxScaler()
mms.fit(x)
```

```python
x = mms.transform(x)
```

```python
pca = PCA(2)
pca.fit(x)
transform = pca.transform(x)
transform
```

```python
svc = svm.SVC(kernel="poly")
svc.fit(transform,y)

DecisionBoundaryDisplay.from_estimator(svc,transform,response_method="predict",cmap=plt.cm.autumn,alpha=0.3)
plt.scatter(x=transform[:,0],y=transform[:,1],c=y)

plt.show()
```
