```python
import pandas as pd
```

```python
df = pd.read_csv("WHO-COVID-19-global-data.csv")
df
```

```python
df["DateReported"] = pd.to_datetime(df["DateReported"])
data = df.groupby("DateReported").sum()
data
```

```python
import plotly.express as px
import plotly.graph_objects as go

figure = go.Figure(data=[
    go.Scatter(name="New Cases",x=data.index,y=data["New_cases"]),
    go.Scatter(name="New Deaths",x=data.index,y=data["New_deaths"])
])
figure.write_html("covid_scatter2.html")
```

```python
import datetime

def friday_13(month,year):
    date = datetime.date(year,month,13)
    if date.weekday() == 4:
        return True
    return False

friday_13(5,2019)
```
