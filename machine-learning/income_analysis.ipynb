```python
import pandas as pd

df = pd.read_csv("adult.csv",header=None)
df.columns = ["age","workclass","fnlwgt","education","educational-num",
    "marital-status","occupation","relationship","race","gender",
    "capital-gain","capital-loss","hours-per-week","native-country","income"]

df["workclass"] = df["workclass"].replace(" ?",None)
df["occupation"] = df["occupation"].replace(" ?",None)
df["native-country"] = df["native-country"].replace(" ?",None)

df = df.dropna()
df
```

```python
df["income"] = df["income"].replace({
    " <=50K": 0,
    " >50K": 1
}).astype("int")
```

```python
from sklearn.preprocessing import LabelEncoder

print(df["education"].value_counts())
le = LabelEncoder()
for column in df.select_dtypes("str").columns:
    df[column] = le.fit_transform(df[column])
    
print(df["education"].value_counts())
```

```python
df.groupby("education")["income"].mean().plot(kind="bar")
```
